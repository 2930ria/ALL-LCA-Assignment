```python
# Student Data Management using Dictionary, Tuple and List

# Tuple
student1 = (101, "Anwesha", "CSE", 92)

# List
student_list = [
    student1,
    (102, "Rahul", "CSE", 85),
    (103, "Priya", "AIML", 89)
]

# Dictionary
students = {
    101: {"Name": "Anwesha", "Branch": "CSE", "Marks": 92},
    102: {"Name": "Rahul", "Branch": "CSE", "Marks": 85},
    103: {"Name": "Priya", "Branch": "AIML", "Marks": 89}
}

# Add a new student
students[104] = {
    "Name": "Amit",
    "Branch": "CSE",
    "Marks": 88
}

# Delete a student
del students[102]

# Update student details
students[103]["Branch"] = "CSE"
students[103]["Marks"] = 94

# Display final records
print("Final Student Records:")

for roll_no, details in students.items():
    print("Roll Number:", roll_no)
    print("Name:", details["Name"])
    print("Branch:", details["Branch"])
    print("Marks:", details["Marks"])
    print()
```
